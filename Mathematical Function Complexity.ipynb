## **How to Rank the Functions**  

When ranking, use these **common Big O categories** as a reference:  

- **Constant Time:** Functions that do not depend on $ n $ (e.g., $ O(1) $).
- **Logarithmic Time:** Functions that grow logarithmically (e.g., $ O(\log n) $).
- **Linear Time:** Functions that grow linearly (e.g., $ O(n) $).
- **Linearithmic Time:** Functions that grow as $ n \log n $.
- **Polynomial Time:** Functions that grow polynomially (e.g., $ O(n^k) $).
- **Exponential Time:** Functions that grow exponentially (e.g., $ O(2^n) $).
- **Factorial Time:** Functions that grow factorially (e.g., $ O(n!) $).

1. **Function:** $f_1(n) = 2$
   - **Big O Complexity:** O(1)
   - **Explanation:** This function has constant time complexity. Its execution time is independent of the input size (n). It always takes the same amount of time, regardless of how large n becomes.

2. **Function:** $f_2(n) = \log(n)$
   - **Big O Complexity:** O(log n)
   - **Explanation:** Logarithmic time complexity. The execution time grows very slowly as the input size increases. For example, if n doubles, the execution time increases by a constant amount.

3. **Function:** $f_{12}(n) = \sqrt{n}$
   - **Big O Complexity:** O(√n)
   - **Explanation:** Square root time complexity. The execution time grows slower than linear time but faster than logarithmic time. It's between O(log n) and O(n).

4. **Function:** $f_3(n) = n$
   - **Big O Complexity:** O(n)
   - **Explanation:** Linear time complexity. The execution time grows directly proportional to the input size. If n doubles, the execution time doubles.

5. **Function:** $f_6(n) = 3n + 5$
   - **Big O Complexity:** O(n)
   - **Explanation:** Linear time complexity. Although there are constants (3 and 5), in Big O notation, we only consider the dominant term as n approaches infinity. The constants and lower-order terms become insignificant.

6. **Function:** $f_4(n) = n \log(n)$
   - **Big O Complexity:** O(n log n)
   - **Explanation:** n log n time complexity. This complexity is slightly worse than linear time but significantly better than quadratic time. It's often seen in efficient sorting algorithms.

7. **Function:** $f_5(n) = n^2$
   - **Big O Complexity:** O(n^2)
   - **Explanation:** Quadratic time complexity. The execution time grows proportionally to the square of the input size. If n doubles, the execution time quadruples.

8. **Function:** $f_7(n) = 4n^2 + 2n + 1$
   - **Big O Complexity:** O(n^2)
   - **Explanation:** Quadratic time complexity. Again, constants and lower-order terms are dropped in Big O notation. The dominant term is n^2.

9. **Function:** $f_8(n) = n^3$
   - **Big O Complexity:** O(n^3)
   - **Explanation:** Cubic time complexity. The execution time grows proportionally to the cube of the input size.

10. **Function:** $f_{11}(n) = 7n^4 + n^2$
    - **Big O Complexity:** O(n^4)
    - **Explanation:** Polynomial time complexity (specifically, quartic). The execution time grows proportionally to the fourth power of the input size.

11. **Function:** $f_9(n) = 5^n$
    - **Big O Complexity:** O(2^n) (or more generally, O(c^n) for any constant c > 1)
    - **Explanation:** Exponential time complexity. The execution time grows *extremely* rapidly as the input size increases. It's often considered computationally infeasible for larger values of n.

12. **Function:** $f_{10}(n) = n!$
    - **Big O Complexity:** O(n!)
    - **Explanation:** Factorial time complexity. This is the slowest growing complexity among the common complexities. The execution time grows *incredibly* fast. Factorial time complexity is generally avoided if possible.
